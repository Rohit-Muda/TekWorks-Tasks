```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('dataset.csv')
df
```

```python
plt.figure(figsize=(10,6))
sns.histplot(x='Age',data=df,kde=True,color='red',edgecolor='black',alpha=0.7)
plt.title("Age Distribution")
plt.show()
```

# 2

```python
#Plot a pie chart showing the percentage distribution of Gender.
plt.figure(figsize=(10,6))
x=df['Gender'].value_counts()
plt.pie(
    x,
    labels=['Male', 'Female'],
    autopct='%1.2f%%',
)

plt.title("Distribution of Target Variable")
plt.show()
```

# 3. Create a bar chart showing the number of customers in each City.


```python
x=df.groupby('City')['Customer_ID'].count()
plt.figure(figsize=(10,6))
sns.barplot(x,color='teal')
```

```python
# 4. Plot a box plot to analyze Monthly Income distribution.
plt.figure(figsize=(6,10))
sns.boxplot(y='Monthly_Income',data=df)
```

```python
# 5. Create a histogram of Spending Score and interpret the skewness.

sns.histplot(x='Spending_Score',data=df,kde=True,color='olive')
plt.title("Spending Score Skewness")
```

```python
# 6. Plot a line chart to visualize Purchase Amount trend over Purchase Date.
sns.lineplot(y=df['Purchase_Amount'],x=df['Purchase_Date'])
plt.
```

```python
# 7. Create a bar plot to compare average Monthly Income by Gender.
x=df.groupby('Gender')['Monthly_Income'].mean()
sns.barplot(x)
```

```python
# 8. Plot a box plot of Purchase Amount grouped by Product Category.
sns.boxplot(x=df['Product_Category'],y=df['Purchase_Amount'])
```
